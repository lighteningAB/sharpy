```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load the CSV file
df = pd.read_csv('runs.csv', header=None, names=['x_position', 'flap_angle'])

# Create time array assuming 100 Hz sampling (i.e., every 0.01 seconds)
dt = 0.01  # seconds
df['time'] = np.arange(len(df)) * dt

# Plotting
plt.figure(figsize=(12, 6))

plt.subplot(2, 1, 1)
plt.plot(df['time'], df['x_position'], label='X Position', color='blue')
plt.ylabel("X Position")
plt.grid(True)
plt.legend()

plt.subplot(2, 1, 2)
plt.plot(df['time'], df['flap_angle'], label='Flap Angle (deg)', color='green')
plt.xlabel("Time (s)")
plt.ylabel("Flap Angle (deg)")
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
