```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load the CSV file
df = pd.read_csv('control15_controlleron_1.csv', header=None, names=['x_position', 'flap_angle'])

# Create time array assuming 100 Hz sampling (i.e., every 0.01 seconds)
dt = 0.01  # seconds
df['time'] = np.arange(len(df)) * dt

# Plotting
plt.figure(figsize=(12, 6))

plt.subplot(2, 1, 1)
plt.plot(df['time'], df['x_position'], label='X Position', color='blue')
plt.ylabel("X Position")
plt.legend()

plt.subplot(2, 1, 2)
plt.plot(df['time'], df['flap_angle'], label='Flap Angle (deg)', color='green')
plt.xlabel("Time (s)")
plt.ylabel("Flap Angle (deg)")
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Load the CSV files
df_on = pd.read_csv('control15_controlleron_2.csv', header=None, names=['x_position', 'flap_angle'])
df_off = pd.read_csv('control15_controlleroff_1.csv', header=None, names=['x_position', 'flap_angle'])

# Create time arrays assuming 100 Hz sampling (i.e., every 0.01 seconds)
dt = 0.01  # seconds
df_on['time'] = np.arange(len(df_on)) * dt
df_off['time'] = np.arange(len(df_off)) * dt

# Plotting overlayed X position
plt.figure(figsize=(12, 6))
plt.plot(df_on['time'], df_on['x_position'], label='Controller ON', color='blue')
plt.plot(df_off['time'], df_off['x_position'], label='Controller OFF', color='red')
plt.xlabel("Time (s)")
plt.ylabel("X Position")
plt.title("X Position Comparison: Controller ON vs OFF")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
