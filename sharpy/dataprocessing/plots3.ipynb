```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file
#0.03 0.00 0.00
#19.65 m/s
# Assuming columns: time_ms, x_position, flap_angle
df = pd.read_csv('control19_on8.csv', header=None, names=['time_ms', 'x_position', 'flap_angle'])

# Normalize time to start at 0 ms
df['time_ms'] = df['time_ms'] - df['time_ms'].iloc[0]

# Plotting
plt.figure(figsize=(12, 6))

plt.subplot(2, 1, 1)
plt.plot(df['time_ms'] / 1000.0, df['x_position'], label='Tip Position', color='blue')
plt.ylabel("Tip Position")

plt.subplot(2, 1, 2)
plt.plot(df['time_ms']/1000.0, df['flap_angle'], label='Flap Angle (deg)', color='green')
plt.xlabel("Time (s)")
plt.ylabel("Flap Angle (deg)")
plt.grid(True)


plt.tight_layout()
plt.show()
```

```python
# Load the CSV files
df_on = pd.read_csv('control19_on8.csv', header=None, names=['time_ms', 'x_position', 'flap_angle'])
df_off = pd.read_csv('control19_off2.csv', header=None, names=['time_ms', 'x_position', 'flap_angle'])

# Create time arrays assuming 100 Hz sampling (i.e., every 0.01 seconds)
dt = 0.01  # seconds
# Normalize time to start at 0 ms
df_on['time_ms'] = df_on['time_ms'] - df_on['time_ms'].iloc[0]
df_off['time_ms'] = df_off['time_ms'] - df_off['time_ms'].iloc[0]


# Plotting
plt.figure(figsize=(12, 6))

# Plotting overlayed X position
plt.plot(df_on['time_ms'] / 1000.0, df_on['x_position'], label='Controller ON', color='blue')
plt.plot(df_off['time_ms'] / 1000.0, df_off['x_position'], label='Controller OFF', color='red')
plt.xlabel("Time (s)")
plt.ylabel("Tip Position")
plt.grid(True)
plt.legend()
plt.xlim(0, max(df_on['time_ms'].max()/1000, df_off['time_ms'].max()/1000))  # <-- Force correct x-axis range
```
